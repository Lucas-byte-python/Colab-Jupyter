```python
# Dicionário de renomeação
novos_nomes = {
    'work_year': 'ano',
    'experience_level': 'senioridade',
    'employment_type': 'contrato',
    'job_title': 'cargo',
    'salary': 'salario',
    'salary_currency': 'moeda',
    'salary_in_usd': 'usd',
    'employee_residence': 'residencia',
    'remote_ratio': 'remoto',
    'company_location': 'empresa',
    'company_size': 'tamanho_empresa'
}

# Aplicando renomeação
df.rename(columns=novos_nomes, inplace=True)

# Verificando resultado
df.columns
```

```python
senioridade = {
    'SE': 'Sênior',
    'MI': 'Pleno',
    'EN': 'Junior',
    'EX': 'Executivo'
}

df['senioridade'] = df['senioridade'].replace(senioridade)
df['senioridade'].value_counts()
```

```python
contrato = {
    'FT': 'Tempo Integral',
    'PT': 'Tempo Parcial',
    'FL': 'Freelance',
    'CT': 'Contrato'
}

df['contrato'] = df['contrato'].replace(contrato)
df['contrato'].value_counts()
```

```python
tamanho_empresa = {
    'S': 'Pequena',
    'M': 'Media',
    'L': 'Grande'
}

df['tamanho_empresa'] = df['tamanho_empresa'].replace(tamanho_empresa)
df['tamanho_empresa'].value_counts()
```

```python
remoto = {
    0: 'Presencial',
    50: 'Híbrido',
    100: 'Remoto'
}

df['remoto'] = df['remoto'].replace(remoto)
df['remoto'].value_counts()
```

```python
import numpy as np

# Criando uma Base de Teste
df_salarios = pd.DataFrame({
    'nome': ['Luiza', 'Lucas', 'Kaio', 'Arthur', 'Val'],
    'salario': [4000, np.nan, 5000, np.nan, 100000]
})

# Calcula a Media, Substitui os Nulos e Arredonda
df_salarios['salario_media'] = df_salarios['salario'].fillna(df_salarios['salario'].mean().round(2))

''' Calcula a Mediana, Substitui os Nulos e Arredonda '''
df_salarios['salario_mediana'] = df_salarios['salario'].fillna(df_salarios['salario'].median())

df_salarios
```

```python
df_temperaturas = pd.DataFrame({
    'Dia': ['Segunda', 'Terça', 'Quarta', 'Quinta', 'Sexta'],
    'Temperatura': [30, np.nan, np.nan, 28, 27]
})

df_temperaturas["preenchido_ffill"] = df_temperaturas["Temperatura"].ffill()
df_temperaturas
```

```python
df_temperaturas = pd.DataFrame({
      'Dia': ['Segunda', 'Terça', 'Quarta', 'Quinta', 'Sexta'],
      'Temperatura': [30, np.nan, np.nan, 28, 27]
})

df_temperaturas["preenchido_bfill"] = df_temperaturas["Temperatura"].bfill()
df_temperaturas
```

```python
df_cidades = pd.DataFrame({
    'nome': ['Luiza', 'Lucas', 'Kaio', 'Arthur', 'Val'],
    'cidade': ['São Paulo', np.nan, 'Curitiba', np.nan, 'Belém']
})

df_cidades['cidade_preenchida'] = df_cidades['cidade'].fillna('Não Informado')
df_cidades
```

```python
import seaborn as sns
```

```python
sns.barplot(data=df_limpo , x='senioridade', y='usd')
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8,5))
sns.barplot(data=df_limpo , x='senioridade', y='usd')
plt.title('Salário Médio por Nível de Senioridade')
plt.xlabel('Senioridade')
plt.ylabel('Salário Médio Anual (USD)')
plt.show()
```

```python
ordem = df_limpo.groupby('senioridade')['usd'].mean().sort_values(ascending=True).index
ordem
```

```python
plt.figure(figsize=(8,5))
sns.barplot(data=df_limpo , x='senioridade', y='usd', order=ordem)
plt.title('Salário Médio por Nível de Senioridade')
plt.xlabel('Senioridade')
plt.ylabel('Salário Médio Anual (USD)')
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.histplot(df_limpo['usd'], bins = 50, kde=True)
plt.title('Dstribuição dos Salário Anuais') # Titulo da Tabela
plt.xlabel('Salário em USD') # Titulo Horizontal
plt.ylabel('Frequência') # Titulo Vertical
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df_limpo['usd'])
plt.title('Boxplot Salário') # Titulo da Tabela
plt.xlabel('Salário em USD') # Titulo Horizontal
plt.show()
```

```python
ordem_senioridade = ['Júnior', 'Pleno', 'Sênior', 'Executivo']

plt.figure(figsize=(8,5))
sns.boxplot(x='senioridade', y='usd', data=df_limpo, order=ordem_senioridade)
plt.title("Boxplot da Distribuição por Senioridade")
plt.xlabel("Salário em USD")
plt.show()
```

```python
ordem_senioridade = ['Júnior', 'Pleno', 'Sênior', 'Executivo']

plt.figure(figsize=(8,5))
sns.boxplot(x='senioridade', y='usd', data=df_limpo, order=ordem_senioridade, palette='Set2', hue='senioridade')
plt.title("Boxplot da Distribuição por Senioridade")
plt.xlabel("Salário em USD")
plt.show()
```

```python
import plotly.express as px
```

```python
df_media_senioridade = df_limpo.groupby('senioridade')['usd'].mean().reindex(ordem_senioridade).reset_index()

fig = px.bar(
    df_media_senioridade,
    x='senioridade',
    y='usd',
    title='Média Salarial por Nível de Senioridade (USD)',
    labels={'senioridade': 'Senioridade', 'usd': 'Salário Médio Anual (USD)'},
    color='senioridade',
    color_discrete_sequence=px.colors.qualitative.Set2
)

fig.update_layout(xaxis={'categoryorder':'array', 'categoryarray':ordem_senioridade})

fig.show()
```

```python
remoto_contagem = df_limpo['remoto'].value_counts().reset_index()
remoto_contagem.columns = ['tipo_trabalho', 'quantidade']

fig = px.pie(
    remoto_contagem,
    names='tipo_trabalho',
    values='quantidade',
    title='Proporção do Tipo de Trabalho',
    hole=0.5
)
fig.update_traces(textinfo='percent+label')
fig.show()
```

Desafio Aula 03: Criar um gráfico de visualização com Plotly com o Salário Médio de Cientista de Dados por País.

```python
df_cientista_dados = df_limpo[df_limpo['cargo'] == 'Data Scientist']

df_salario_cientista_por_pais = df_cientista_dados.groupby('residencia')['usd'].mean().reset_index()

# Criando o gráfico de barras interativo com Plotly
fig = px.bar(
    df_salario_cientista_por_pais,
    x='residencia',
    y='usd',
    title='Salário Médio de Cientista de Dados por País (USD)',
    labels={'residencia': 'País', 'usd': 'Salário Médio Anual (USD)'},
    color='residencia',
    color_discrete_sequence=px.colors.qualitative.Set3
)

fig.update_layout(xaxis_title='País de Residência', yaxis_title='Salário Médio Anual (USD)')
fig.show()
```
